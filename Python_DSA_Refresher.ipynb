```python
#Calculates the number of seconds in day
def dayCalc():
    #input days from user
    days = int(input("Enter a number of days: "))
    #calculate seconds from days
    seconds = days * 24*3600
    #print the result
    print(f"{days} days is {seconds} seconds of time.")

dayCalc()
```

```python
#Find volume of a sphere

def sphereVolume():
    #User input
    radius = float(input("Enter the radius of this sphere:"))
    #Volume calculation/Definition of pi
    pi = 3.14159
    volume = (4/3) * pi * radius**3
    #Printing the result
    print(f"The volume of sphere, radius {radius} is {volume}.")
    
sphereVolume()
```

```python
#perimeter and area of a square

def squarePerimeter():
    #user input
    side_length = float(input("Enter the length of one side of the square: "))
    #perimeter and area calculation happening here
    perimeter = 4 * side_length
    area = side_length ** 2
    #printing the results
    print(f"The perimeter of the square is: {perimeter}")
    print(f"The area of the square is: {area}")

squarePerimeter()
```

```python
#Checking whether an input character is uppwer or lower case.

def upperorLower():
    input_letter = input("Enter a character:")
    #test condition for uppercase and lowercase letters
    if input_letter.isupper():
        print(f"{input_letter} is uppercase")
    elif input_letter.islower():
        print(f"{input_letter} is lowercase")
        #case for if the input is not a letter
    else:
        print(f"{input_letter} is not a letter")

upperorLower()
```

```python
#Manipulation of the x and y values

def XandY():
    x = 0
    y = 20
    while y > 6:
        y -= 4
        x += 2/y
    print(f"The final value of x is: {x}")
    
XandY()
```

```python
#Finding the total and average of 5 numbers put into an array

def loopinginput():
    i = 0
    input_array = []
    while i < 5:
        user_input = input("Enter a number: ")
        input_array.append(user_input)
        i += 1
    
    total = sum(map(float, input_array))
    avg = total / len(input_array)
    print(f"The total of the numbers you entered is: {total}")
    print(f"The average of the numbers you entered is: {avg}")
        
loopinginput()
```
